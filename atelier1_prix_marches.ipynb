```python
# 1. Quelle période couvrent les relevés ?
date_debut = releves["date"].min()
date_fin = releves["date"].max()
print(f"date_debut : {date_debut} | date_fin: {date_fin}")
```

**La periode qui couvre les relevés s'etend du 2025-01-15 à 2026-07-15**

```python
#2Combien de marchés différents ? Combien de régions ? Lesquelles (triées avec sorted)
print(releves["market"].nunique())
print(releves["admin1"].nunique())
print(sorted(releves["market"].unique()))
sorted(releves["market"].unique())
```

**Il y'a 67 marchés differents**
**ET 14 regions**

```python
#4). Combien de monnaies, d'unités et de types de prix différents ? Lesquels ?
print(releves["currency"].nunique())
print(releves["unit"].nunique())
print(releves["pricetype"].nunique())
print(f"Monaies:{sorted(releves["currency"].unique())}")
print(f"Unités:{sorted(releves["unit"].unique())}")
print(f"Type_prix:{sorted(releves["pricetype"].unique())}")
```

**Il y'a qu'une seule monaie(XOF), deux unités(Head, KG) et un type-prix(Retail)**

```python
#Une unité est Head (« par tête ») : à quelle sorte de denrée correspond-elle,
print(f"Il correspond :{releves[releves["unit"] == "Head"]["commodity"].unique()}")
```
